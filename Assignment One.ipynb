```python
# ==========================================
# Mini DSA Project: Student Performance Analyzer
# ==========================================

def enter_marks(marks):
    """Allows user to enter multiple student marks into the list."""
    try:
        count = int(input("How many marks would you like to enter? "))
        for i in range(count):
            while True:
                try:
                    mark = float(input(f"Enter mark for student {i + 1} (0-100): "))
                    if 0 <= mark <= 100:
                        marks.append(mark)
                        break
                    else:
                        print("Invalid mark! Please enter a value between 0 and 100.")
                except ValueError:
                    print("Invalid input! Please enter a numeric value.")
        print(f"Successfully added {count} marks.\n")
    except ValueError:
        print("Invalid number entered.\n")

def display_marks(marks):
    """Displays all current marks stored in the collection."""
    if not marks:
        print("No marks entered yet.\n")
        return
    print("\n--- Current Student Marks ---")
    for idx, mark in enumerate(marks, start=1):
        print(f"Student {idx}: {mark}")
    print()

def find_min_max(marks):
    """Determines highest and lowest marks manually without max()/min()."""
    if not marks:
        print("No marks available to analyze.\n")
        return
    
    highest = marks[0]
    lowest = marks[0]
    
    for mark in marks:
        if mark > highest:
            highest = mark
        if mark < lowest:
            lowest = mark
            
    print(f"\n Highest Mark: {highest}")
    print(f" Lowest Mark: {lowest}\n")

def linear_search(marks, target):
    """Custom Linear Search implementation (no built-in index()/in)."""
    if not marks:
        print("No marks available to search.\n")
        return
    
    found_indices = []
    for i in range(len(marks)):
        if marks[i] == target:
            found_indices.append(i)
            
    if found_indices:
        print(f"\n Mark {target} found at index positions: {found_indices}\n")
    else:
        print(f"\n Mark {target} was not found in the record.\n")

def bubble_sort(marks):
    """Custom Bubble Sort implementation (no built-in sort()/sorted())."""
    if not marks:
        print("No marks available to sort.\n")
        return
    
    n = len(marks)
    sorted_marks = marks.copy()
    
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if sorted_marks[j] > sorted_marks[j + 1]:
                sorted_marks[j], sorted_marks[j + 1] = sorted_marks[j + 1], sorted_marks[j]
                swapped = True
        if not swapped:
            break
            
    print("\n--- Marks Sorted in Ascending Order ---")
    print(sorted_marks)
    print()
    return sorted_marks

def main_menu():
    """Main Interactive Menu System."""
    marks_list = []
    
    while True:
        print("========================================")
        print("    STUDENT PERFORMANCE ANALYZER        ")
        print("========================================")
        print("1. Accept Student Marks")
        print("2. Display Marks Entered")
        print("3. Find Highest and Lowest Marks")
        print("4. Search for a Mark")
        print("5. Sort Marks (Ascending Order)")
        print("6. Exit")
        print("========================================")
        
        choice = input("Select an option (1-6): ").strip()
        
        if choice == '1':
            enter_marks(marks_list)
        elif choice == '2':
            display_marks(marks_list)
        elif choice == '3':
            find_min_max(marks_list)
        elif choice == '4':
            try:
                target_val = float(input("Enter the mark to search for: "))
                linear_search(marks_list, target_val)
            except ValueError:
                print("Please enter a valid numeric value.\n")
        elif choice == '5':
            bubble_sort(marks_list)
        elif choice == '6':
            print("Exiting program. Goodbye!")
            break
        else:
            print("Invalid selection. Please choose a valid menu option (1-6).\n")

# --- EXECUTION WITH 10 TEST MARKS ---
test_marks = [75, 95, 69, 85, 56, 64, 89, 36, 43, 71]

print("--- AUTOMATED TEST DEMO (10 Student Marks) ---")
print(f"Test Data: {test_marks}")

print("\n1. Finding Min/Max:")
find_min_max(test_marks)

print("2. Searching for Mark 45.0:")
linear_search(test_marks, 45.0)

print("3. Sorting Marks:")
bubble_sort(test_marks)
```

## Section 3: Data Structures & Algorithms Explanation

### 1. Data Structure Used
* **Data Structure:** Dynamic Array (`list`)
* **Explanation:** Python lists store elements sequentially in contiguous memory blocks and dynamically resize as elements are added. They support fast $O(1)$ constant-time positional access and sequential iteration, making them ideal for managing student score lists.

### 2. Searching Algorithm Used
* **Algorithm:** Linear Search
* **Explanation:** Linear Search checks each item sequentially from index `0` to `N-1`. It compares every element against the target value until match positions are identified or all items are examined. This avoids built-in search functions like `in` or `.index()`.

### 3. Sorting Algorithm Used
* **Algorithm:** Bubble Sort
* **Explanation:** Bubble Sort iteratively steps through the array, comparing neighboring elements and swapping them if they are out of order (i.e., `list[j] > list[j+1]`). With each pass, the next largest element "bubbles up" to its correct position. This avoids built-in methods like `.sort()` or `sorted()`.

### 4. Big-O Time Complexity
* **Complexity:** $\mathcal{O}(N^2)$ Quadratic Time Complexity for Bubble Sort.
* **Explanation:** In the worst-case scenario (an array sorted in reverse order), Bubble Sort performs $\frac{N(N-1)}{2}$ comparison and swap operations for $N$ records. Consequently, running time grows quadratically relative to input size.
